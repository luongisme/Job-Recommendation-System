```python
import pandas as pd
import numpy as np
import psycopg2.extras as extras
```

### Load Configs ###

```python
import sys
sys.path.append("..")

from config import get_connection, logger
```

### Company Industries ###

```python
logger.info("Loading company_industries")

df = pd.read_csv("../raw_data/company_industries.csv")
df.columns = ["industry", "company_id"]
df["company_id"] = df["company_id"].astype(int)

query = """
INSERT INTO company_industries (industry, company_id)
VALUES %s
ON CONFLICT DO NOTHING
"""

conn = get_connection()
with conn.cursor() as cur:
    extras.execute_values(cur, query, df.values.tolist())
    conn.commit()
conn.close()
```

### Employee Counts ###

```python
logger.info("Loading employee_counts")

df = pd.read_csv("../raw_data/employee_counts.csv")
df.columns = ["employee_count", "follower_count", "time_recorded", "company_id"]
df[["employee_count", "follower_count"]] = df[["employee_count", "follower_count"]].apply(pd.to_numeric, errors = "coerce")

query = """
INSERT INTO employee_counts (
    employee_count, follower_count, time_recorded, company_id
)
VALUES %s
"""

conn = get_connection()
with conn.cursor() as cur:
    extras.execute_values(cur, query, df.values.tolist())
    conn.commit()
conn.close()
```
