```python
total_cell=np.prod(df.shape)
missing_value=df.isnull().sum()
missing_percentage = (missing_value.sum() / total_cell) * 100
print(f"Total missing values: {missing_value.sum()}")
print(f"Missing percentage: {missing_percentage:.2f}%")
```

```python
#cannot identify the logic to fill in company name so drop null rows
df = df.dropna(subset=['company_name'])
```

```python
#these col missing values should be 0 as there was nothing to record
df['applies']=df['applies'].fillna(0).astype(int)
df['remote_allowed']=df['remote_allowed'].fillna(0).astype(int)
df['views']=df['views'].fillna(0).astype(int)
```

```python
#drop sponsored as it only have 1 value and doesnt contribute any value to analyze problem
df=df.drop(columns='sponsored')

#missing 95%, cant help it
df=df.drop(columns='med_salary')

#no analysis value and high missing percentage
df=df.drop(columns='closed_time')

#missing 98%, the data could be foreign key, not real data
df=df.drop(columns="skills_desc")

#compensation_type have the same role as pay period but pay_period specify the info better
df=df.drop(columns='compensation_type')
```

```python

url_cols = [
    "job_posting_url",
    "application_url",
    "posting_domain"
]

df_urls = df[["job_id"] + url_cols].copy()
df_clean = df.drop(columns=url_cols)
df_clean.to_csv("jobs_clean_features.csv", index=False)
df_urls.to_csv("jobs_urls.csv", index=False)
```
